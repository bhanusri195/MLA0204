```python

import numpy as np

np.random.seed(42)

X = np.concatenate([
    np.random.normal(5, 1, 100),
    np.random.normal(10, 1, 100)
])

means = np.random.choice(X, 2)
variances = np.ones(2)
weights = np.ones(2) / 2

for i in range(100):
    p = np.zeros((len(X), 2))

    for j in range(2):
        p[:, j] = weights[j] * np.exp(
            -0.5 * (X - means[j])**2 / variances[j]
        ) / np.sqrt(2 * np.pi * variances[j])

    p /= p.sum(axis=1, keepdims=True)

    for j in range(2):
        r = p[:, j]
        total = r.sum()
        means[j] = np.sum(r * X) / total
        variances[j] = np.sum(r * (X - means[j])**2) / total
        weights[j] = total / len(X)

print("Final Means:", means)
print("Final Variances:", variances)
print("Final Weights:", weights)
```
